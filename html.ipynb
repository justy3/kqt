## HTML table rendering checks
Run the load cell above first, then each example below. `displayhtmltab[0;0;table]` renders every row and column inside the configured scroll height (420 px by default), independently of `\c`.


### 1. Vertical scrolling: 1,000 rows
Scroll down to row 999 and check that the column headers stay visible.


```q
verticalTable:([] row:til 1000; sym:1000#`AAPL`MSFT`GOOG`AMZN; qty:1+til 1000; price:100+0.01*til 1000);
displayhtmltab[0;0;verticalTable]
```

### 2. Horizontal scrolling: 80 columns
Scroll right to `column_80`. Each cell contains its column number, making alignment easy to check.


```q
wideColumns:`$("column_",/:string 1+til 80);
wideTable:flip wideColumns!{20#x} each 1+til 80;
displayhtmltab[0;0;wideTable]
```

### 3. Key rendering: two key columns, 200 rows
Check that `sym` and `tradeDate` have distinct key styling in both the header and body. Scroll vertically to check that styling persists.


```q
keyedTable:`sym`tradeDate xkey ([] sym:200#`AAPL`MSFT`GOOG`AMZN; tradeDate:2026.01.01+"j"$floor (til 200)%4; qty:10*1+til 200; price:100+0.25*til 200);
displayhtmltab[0;0;keyedTable]
```
